```python
import tensorflow as tf
```

```python
mnist_data=tf.keras.datasets.fashion_mnist
```

```python
(train_images,train_labels),(test_images,test_labels)=mnist_data.load_data()
```

```python
#random sample from datasets
print(train_images[5],train_labels[5])
```

```python
#plot the train image

import matplotlib.pyplot as plt
```

```python
plt.imshow(train_images[5])
plt.show()
```

```python
# normalization
train_images=train_images/255.0
test_images=test_images/255.0
```

```python
print(train_images[0])
```

```python
#create a deep learning model
model=tf.keras.Sequential([
    tf.keras.Input(shape=(28,28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(512,activation="relu"),
    tf.keras.layers.Dense(256,activation="relu"),
    tf.keras.layers.Dense(128,activation="relu"),
    tf.keras.layers.Dense(64,activation="relu"),
    tf.keras.layers.Dense(32,activation="relu"),
    tf.keras.layers.Dense(10,activation="softmax"),

])
```

```python
model.compile(optimizer="Adam",loss="sparse_categorical_crossentropy", metrics=['accuracy'])
```

```python
model.fit(train_images,train_labels,epochs=8)
```

```python
model.evaluate(test_images,test_labels)
```

```python

predicted_labels=model.predict(test_images)
```

```python
print(len(predicted_labels))
```

```python
# print(test_images[5])
import numpy as np
plt.imshow(test_images[5])
print(test_labels[5]) 
print(f"Predicted outputs={np.argmax(predicted_labels[5])} and actual output={test_labels[5]}")
```

```python
y_pred=np.zeros(len(predicted_labels))
for i in range(len(predicted_labels)):
    y_pred[i]=np.argmax(predicted_labels[i])
```

```python
# confusion matrix
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
cm=confusion_matrix(test_labels,y_pred)
```

```python
cr=classification_report(test_labels,y_pred)
print(cr)
```

```python
print(accuracy_score(test_labels,y_pred))
```

```python
# visulaize confusion matrix
import seaborn as sns
```

```python
sns.heatmap(cm,cmap='plasma',annot=True, xticklabels=[0, 10],yticklabels=[0, 10])
plt.xlabel("Actual Output")
plt.ylabel("Predicted Output")
plt.title("Fashion MNSIT conf matrix")
plt.tight_layout()
plt.show()
```
